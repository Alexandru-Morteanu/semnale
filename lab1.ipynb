a) In Python, simulat, i axa real˘a de timp printr-un s, ir de numere suficient de apropiate, spre exemplu [0 : 0.0005 : 0.03].

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0.0, 0.03, 0.0005)

x = np.cos(520 * np.pi * t + np.pi/3)
y = np.cos(280 * np.pi * t - np.pi/3)
z = np.cos(120 * np.pi * t + np.pi/3)

fig, axs = plt.subplots(3)
fig.suptitle("Primul exercițiu")
axs[0].plot(t, x)
axs[1].plot(t, y)
axs[2].plot(t, z)
```

```python
perioada_de_esantionare = 0.0005 * 10
t = np.arange(0.0, 0.03, perioada_de_esantionare)

x = np.cos(520 * np.pi * t + np.pi/3)
y = np.cos(280 * np.pi * t - np.pi/3)
z = np.cos(120 * np.pi * t + np.pi/3)

fig, axs = plt.subplots(3)
fig.suptitle("Primul exercițiu")

axs[0].plot(t, x)
axs[1].plot(t, y)
axs[2].plot(t, z)
```
